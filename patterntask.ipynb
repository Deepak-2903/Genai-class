```python
num = int(input("Enter a num:"))
symbol = input("Enter a symbol:")
for x in range(num,0,-1):
    for y in range(x):
        print(symbol, end=" ")
    print()
```

```python
# print below pattern
'''
&&&&
&&&
&&
&
'''
num = int(input("Enter a num: "))
symbol = input("Enter a symbol: ")
x = num
while x > 0:
    y = 0
    while y < x:
        print(symbol, end=" ")
        y += 1
    print()
    x -= 1
```

```python
# print below pattern
# enter a number between 1 to 9
'''
1
22
333
4444
55555
666666
7777777
88888888
999999999
'''
n = int(input("Enter a number between 1 to 9: "))
for i in range(1, n + 1):
    for j in range(i):
        print(i, end="")
    print()
```
