3. Explain type casting with two examples.

Type casting is the process of converting a value from one data type to another.

Example 1:
a = "10"
b = int(a)
print(b)
Output: 10

Example 2:
x = 10
y = float(x)
print(y)
Output: 10.0

4. What is the difference between a list and a tuple?

List: List is mutable, written using [], Elements can be changed.
Tuple: Tuple is immutable, Written using(), Elements cannot be changed.
Example:
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

5. What is the difference between a set and a dictionary?

Set: Stores unique values, written using{} with values.
Dictionary: Stores key-values pairs, written using{} with key-values pairs.
Example:
my_set = {10, 20, 30}
my_dict = {"name": "Arun", "age": 20}

```python
### 1. Student Details 
### Get student name, age and course using input(). Convert age to int and display the details using print().

name = input("Enter student name: ")
age = int(input("Enter age: "))
course = input("Enter course: ")
print("Student Name:", name)
print("Age:", age)
print("Course:", course)
```

```python
### 2. Simple Calculator 
### Get two numbers from the user and display addition, subtraction, multiplication, division, floor division and modulus. 

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Floor Division:", num1 // num2)
print("Modulus:", num1 % num2)
```
