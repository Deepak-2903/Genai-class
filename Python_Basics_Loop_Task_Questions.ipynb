```python
# 24.Write a program to create a simple calculator using if-elif-else. 
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")
if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "*":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid operator")
```

```python
#29.Write a program using while loop to print all odd numbers from 1 to 50. 
i = 1
while i <= 50:
    if i % 2 != 0:
        print(i)
    i += 1
```

```python
#30.Write a program using while loop to calculate the sum of numbers from 1 to 100. 
i = 1
total = 0
while i <= 100:
    total += i
    i += 1
print("Sum:", total)
```

```python
#31.Write a program using while loop to display the multiplication table of a given number. 
num = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(num, "x", i, "=", num * i)
    i += 1
```

```python
#32.Write a program using while loop to find the factorial of a number. 
num = int(input("Enter a number: "))
factorial = 1
i = 1
while i <= num:
    factorial *= i
    i += 1
print("Factorial:", factorial)
```

```python
#33.Write a program using while loop to count the number of digits in a number. 
num = int(input("Enter a number: "))
num = abs(num)
if num == 0:
    count = 1
else:
    count = 0
    while num > 0:
        num //= 10
        count += 1
print("Number of digits:", count)
```

```python
#34.Write a program using while loop to reverse a given number. 
num = int(input("Enter a number: "))
original = num
reverse = 0
num = abs(num)
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10
if original < 0:
    reverse = -reverse
print("Reversed number:", reverse)
```

```python
#35.Write a program using while loop to check whether a number is a palindrome. 
num = int(input("Enter a number: "))
original = num
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10
if original == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
#41.Write a program using for loop to find the factorial of a number. 
num = int(input("Enter a number: "))
factorial = 1
for i in range(1, num + 1):
    factorial *= i
print("Factorial:", factorial)
```

```python
#42.Write a program using for loop to count the vowels in a given string. 
text = input("Enter a string: ")
count = 0
for character in text:
    if character.lower() in "aeiou":
        count += 1
print("Number of vowels:", count)
```
