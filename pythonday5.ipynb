```python
score = 100
secret_number = 50
guess = int(input("Enter your guess:"))

if guess == secret_number:
    print("correct")
    score = score + 20
    print("Your score:", score)
elif guess > secret_number:
    print("Too high")
    score = score - 10
    print("Your score:", score)
else:
    print("Too low")
    score = score - 10
    print("Your score:", score)
```
